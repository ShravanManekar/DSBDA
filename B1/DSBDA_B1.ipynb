```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("dataset_Facebook.csv",sep=";")
```

```python
print("First 5 Rows:")
df.head()
```

```python
subset1 = df[['Type', 'Category', 'Paid']]

print("\nSubset with selected columns:")
subset1.head()
```

```python
subset2 = df[df['Paid'] == 1]

print("\nSubset where Paid = 1:")
subset2.head()
```

```python
# Create first dataframe
data1 = {
    'Post ID': [1, 2, 3, 4],
    'Likes': [100, 200, 300, 400]
}

df1 = pd.DataFrame(data1)

# Create second dataframe
data2 = {
    'Post ID': [1, 2, 3, 4],
    'Comments': [10, 20, 30, 40]
}


```

```python
df2 = pd.DataFrame(data2)
merged_df = pd.merge(df1, df2, on='Post ID')

print("\nMerged Data:")
merged_df
```

```python
print("\n========== SORT DATA ==========")

# Sort dataset by Total Interactions
sorted_df = df.sort_values(by='Total Interactions',ascending=False)

print("\nData sorted by Total Interactions:")
sorted_df[['Type', 'Total Interactions']].head()
```

```python
print("\n========== TRANSPOSE DATA ==========")

transpose_df = df.head().T

print("\nTransposed Data:")
transpose_df
```

```python
print("\n========== RESHAPED DATA ==========")

reshaped_df = pd.melt(df,id_vars=['Type'],value_vars=['comment', 'like', 'share'])

print("\nReshaped Data:")
reshaped_df.head()
```
