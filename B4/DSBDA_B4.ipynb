```python
import numpy as np
import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
air_df = pd.read_csv("air_quality.csv")
heart_df = pd.read_csv("heart.csv")
```

```python
air_df.head()
```

```python
heart_df.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(air_df['AQI'],bins=10,edgecolor='black')

plt.title("AQI Distribution")
plt.xlabel("AQI")
plt.ylabel("Frequency")

plt.show()
```

```python
# Countplot using seaborn
plt.figure(figsize=(6,5))

sns.countplot(x='target', data=heart_df)

plt.title("Heart Disease Count")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(heart_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(heart_df['age'],heart_df['chol'],alpha=0.7)

plt.title("Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(air_df.iloc[:, 0],bins=20,edgecolor='black')

plt.title("Air Quality Data Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```
